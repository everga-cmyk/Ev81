# Sistema Solare Animato con REBOUND

100 anni di orbite planetarie in movimento!

```python
!pip install -q rebound matplotlib numpy --quiet
import rebound
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
%matplotlib inline
```

```python
sim = rebound.Simulation()
sim.units = ('AU', 'yr', 'Msun')
sim.add(m=1)
sim.add(m=1.66e-7, a=0.387, e=0.206)
sim.add(m=2.45e-6, a=0.723, e=0.007)
sim.add(m=3e-6, a=1, e=0.017)
sim.add(m=3.23e-7, a=1.524, e=0.093)
sim.add(m=9.55e-4, a=5.203, e=0.048)
sim.add(m=2.86e-4, a=9.537, e=0.056)
sim.add(m=4.37e-5, a=19.191, e=0.046)
sim.add(m=5.15e-5, a=30.069, e=0.010)
sim.move_to_com()
sim.integrator = "ias15"

colors = ['#FDB813','#B8860B','#FFC649','#4169E1','#CD5C5C','#DAA520','#F4A460','#4FD0E0','#4169E1']
sizes = [200,30,50,50,40,140,110,80,80]
names = ['Sole','Mercurio','Venere','Terra','Marte','Giove','Saturno','Urano','Nettuno']
```

```python
fig, ax = plt.subplots(figsize=(12,12))
ax.set_facecolor('black')
ax.set_xlim(-35,35)
ax.set_ylim(-35,35)
ax.set_aspect('equal')
ax.axis('off')

points = [ax.scatter([], [], s=sizes[i], color=colors[i], edgecolor='white', zorder=5) for i in range(9)]
trails = [ax.plot([], [], color=colors[i], lw=1, alpha=0.7)[0] for i in range(9)]
title = ax.text(0.5, 0.95, '', transform=ax.transAxes, ha='center', color='white', fontsize=18, fontweight='bold')

def animate(t):
    sim.integrate(t)
    title.set_text(f'Sistema Solare — Anno {t:.1f}')
    for i, p in enumerate(sim.particles):
        points[i].set_offsets([p.x, p.y])
        trail_x = trails[i].get_xdata().tolist() + [p.x]
        trail_y = trails[i].get_ydata().tolist() + [p.y]
        if len(trail_x) > 200: 
            trail_x = trail_x[-200:]
            trail_y = trail_y[-200:]
        trails[i].set_data(trail_x, trail_y)
    return points + trails

ani = FuncAnimation(fig, animate, frames=np.linspace(0, 100, 2000), interval=30, blit=False)
HTML(ani.to_jshtml())
```
